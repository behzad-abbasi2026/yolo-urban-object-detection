- Python
- Jupyter Notebook
- Ultralytics YOLO


## 5. Loading the Pretrained YOLO Model

```python
# Import the YOLO class
from ultralytics import YOLO

# Load the pretrained YOLO model
model = YOLO("yolo11n.pt")
```

## 6. Image Input and Preprocessing

 **Select an urban image in JPG, JPEG, PNG, or BMP format for object detection.**

This section prepares user-provided images for object detection. It checks the input image and handles image format conversion and resizing when necessary.

```python
from tkinter import Tk, filedialog
from pathlib import Path

# Open a file selection window
root = Tk()
root.withdraw()
root.attributes("-topmost", True)

image_path = filedialog.askopenfilename(
    title="Select an urban image",
    filetypes=[
        ("Image files", "*.jpg *.jpeg *.png *.bmp"),
        ("All files", "*.*")
    ]
)

root.destroy()

if image_path:
    image_path = Path(image_path)
    print("Selected image:", image_path)
else:
    print("No image selected.")
```

# 7. Run YOLO object detection on the selected image

```python

results = model.predict(source=str(image_path), save=True)

# Display the detection results
results[0].show()
```

# 8. Extract detected object information


```python
boxes = results[0].boxes

# Count detected objects
print("Number of detected objects:", len(boxes))

# Display detected classes and confidence scores
for box in boxes:
    class_id = int(box.cls[0])
    confidence = float(box.conf[0])
    class_name = model.names[class_id]

    print(f"Object: {class_name}, Confidence: {confidence:.2f}")
```

## 9. Conclusion

This project demonstrates how a pretrained YOLO model can detect objects in urban images. Users can select their own images, run object detection, visualize the results, and examine the detected object classes and confidence scores.

## 10. Limitations

- The model uses pretrained weights and has not been fine-tuned specifically for urban resilience applications.
- Detection performance depends on image quality, object visibility, and the model's pretrained classes.
- Confidence scores indicate the model's confidence in its predictions and do not guarantee that every detection is correct.
- The project has been tested on selected images and has not been evaluated on a large, systematically collected dataset.
